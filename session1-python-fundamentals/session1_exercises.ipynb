<details>
<summary>A <b>list comprehension</b> is a compact way to create a new list by iterating over an iterable and optionally transforming or filtering its elements.</summary>

The basic structure is:
```python
[expression for item in iterable]
```
For example, given a list of numbers you would like to square, you could do

numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in numbers if x % 2 == 0]

<details>
<summary><code>enumerate</code> returns an iterator that yields tuples containing each element's index and value.</summary>
<code>enumerate()</code> produces pairs of:

```python
fruits = ["apple", "banana", "cherry"]

<code>
(0, 'apple')

for index, value in enumerate(fruits):

names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

for item in zip(names, ages):
    print(item)
```
This will print:

("Alice", 25)
("Bob", 30)
("Charlie", 35)

You could directly unpack the <code>item</code> using the pattern seen with <code>enumerate</code>.

### Exercise 2.8 (core, demo) — The mutable default argument trap

Run the cell below **as-is** first, and look closely at the output. This is one of Python's most famous "gotchas" — worth knowing about *before* it bites you in a real project.


````python
def add_item(item, cart=[]):  # <-- the trap is right here
    cart.append(item)
    return cart

print(add_item("apple"))
print(add_item("banana"))   # you might expect ["banana"] ... but you won't get that
````

**What happened?** Default argument values are evaluated **once**, when the function is defined — not each time it's called. So `cart=[]` creates a single list that's reused (and mutated!) across every call that doesn't pass its own `cart`.

**The fix**: use `None` as the default, and create the mutable object inside the function.

    loop            loop
